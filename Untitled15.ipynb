1. **department – Fill using Mode**

Mode means the most frequently occurring value.

```python
df["department"] = df["department"].fillna(
    df["department"].mode()[0]
)
```

2. **work_accident – Fill using mode**

```python
df["work_accident"] = df["work_accident"].fillna(
    df["work_accident"].mode()[0]
)
```

**3. emp_id – Fill using Unknown**

Employee IDs are unique identifiers. We should not guess missing IDs.

```python
df["emp_id"] = df["emp_id"].fillna("Unknown")
```

**4. left_company – Keep missing values**

This is the target column.
- 0 = Employee stayed.
- 1 = Employee left.
Do not fill missing values with mean or mode because that would create incorrect labels

**5. salary_k – Fill using Median**

Median is the middle value. It is useful when numerical data contains outliers.

```python
df["salary_k"] = df["salary_k"].fillna(
    df["salary_k"].median()
)
```

**6. education_level – Fill using Mode**

```python
df["education_level"] = df["education_level"].fillna(
    df["education_level"].mode()[0]
)
```

**7. promotion_last_5years – Fill using Mode**

```python
df["promotion_last_5years"] = df["promotion_last_5years"].fillna(
    df["promotion_last_5years"].mode()[0]
)
```

**8. years_at_company – Fill using Median**

```python
df["years_at_company"] = df["years_at_company"].fillna(
    df["years_at_company"].median()
)
```

**9. salary_k_dup – Remove duplicate column**

This column contains the same values as salary_k.

```python
df.drop("salary_k_dup", axis=1, inplace=True)
```

**10. last_performance_score – Fill using Median**

```python
df["last_performance_score"] = df["last_performance_score"].fillna(
    df["last_performance_score"].median()
)
```

**checking missing value**

```python
df = df.dropna(subset=["left_company"])
```

**missing value in left_company**

```python

df["left_company"].isnull().sum()
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.info()
```

**suitable datatype for column**

```python
df['department'] = df['department'].astype('category')

df['work_accident'] = df['work_accident'].astype('category')

df['emp_id'] = df['emp_id'].astype('string')

df['left_company'] = df['left_company'].astype('Int64')

df['salary_k'] = df['salary_k'].astype('float64')

df['education_level'] = df['education_level'].astype('category')

df['promotion_last_5years'] = df['promotion_last_5years'].astype('category')


df['years_at_company'] = df['years_at_company'].round().astype('Int64')

df['last_performance_score'] = df['last_performance_score'].round().astype('Int64')

df.info()
```

```python
df.info()
```

```python
df['department'] = df['department'].str.strip().str.lower()

df['department'] = df['department'].replace({
    'sals': 'sales',
    'saes': 'sales',
    'ales': 'sales',
    'finace': 'finance',
    'inance': 'finance',
    'suppot': 'support',
    'suport': 'support',
    'engineerig': 'engineering',
    'engneering': 'engineering'
})

df['department'] = df['department'].str.title()

df['work_accident'] = df['work_accident'].replace({
    'NO': 'No', 'N': 'No', 'o': 'No'
})

df['promotion_last_5years'] = df['promotion_last_5years'].replace({
    'NO': 'No', 'N': 'No', 'o': 'No',
    'YES': 'Yes', 'es': 'Yes', 'Ye': 'Yes'
})
```

**Bar plot – Employees by department**

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='department')

plt.title('Employees by Department')
plt.xlabel('Department')
plt.ylabel('Employee Count')
plt.xticks(rotation=45)
plt.show()
```

**Count plot – Employee attrition**

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='left_company')

plt.title('Employee Attrition')
plt.xlabel('Left Company (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.show()
```

**. Histogram – Salary distribution**

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x='salary_k', bins=15, kde=True)

plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```

**6. Histogram – Years at company**

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x='years_at_company', bins=15, kde=True)

plt.title('Years at Company Distribution')
plt.xlabel('Years')
plt.ylabel('Employee Count')
plt.show()
```

 **Box plot – Salary vs attrition**

```python
plt.figure(figsize=(7,5))

sns.boxplot(data=df, x='left_company', y='salary_k')

plt.title('Salary vs Employee Attrition')
plt.xlabel('Left Company (0 = No, 1 = Yes)')
plt.ylabel('Salary')
plt.show()
```

** Heatmap – Correlation**

```python
plt.figure(figsize=(8,6))

corr = df.select_dtypes(include='number').drop(
    columns=['salary_k_dup'], errors='ignore'
).corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```
